# Derive Boundary and Turbine Layout From Sea-Impact

This Jupiter Notebook will show how to get a boundary and turbine layout of a site, sourced from Sea Impact. Make sure you have the following require files in the same directory:

* `layout_dev.py`
* `boundary_dev.py`
* The `.geojson` files

## Boundary

Provide the `filepath`

```python
# Replace here
# You can get the full filepath by right click the .geojson file in the Explorer Panel and press `copy path` then add it below
boundary_filepath = "/Users/Antonio/Desktop/Research_Projects/WESL/optimizer/data/DigitizeLayers/US/Vineyard.geojson"
```

Extract Boundary using function `get_only_boundary` from `boundary_dev.py`

```python
from boundary_dev import get_only_boundary

eastings, northings = get_only_boundary(boundary_filepath)
```

Print the boundaries:

```python
print(f'eastings = {eastings}\nnorthings = {northings}')
```

Now we plot the boundary as following:

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(eastings, northings, 'k', lw=2.5)
plt.tight_layout()
plt.show()
```

## Turbine Layout

**Note:** For the turbine layout the geojson file must be a line in which the points are the position of the turbines, if you provide the boundary file instead, the turbines layout will be turbines positioned at the edges of the boundary, keep that in mind.

```python
# Replace here
layout_filepath = "/Users/Antonio/Desktop/Research_Projects/WESL/optimizer/vineyardwind_TBL.geojson"
```

```python
from layout_dev import SeaImpactTurbineLayout

pos_x, pos_y = SeaImpactTurbineLayout(layout_filepath)
```

```python
print(f'wt_x = {pos_x} \nwt_y = {pos_y}')
```

```python
plt.figure()
plt.scatter(pos_x, pos_y, marker='2', color='k')
plt.tight_layout()
plt.show()
```

## Boundary + Layout 

```python
plt.figure()
plt.plot(eastings, northings, 'k', lw=2)
plt.scatter(pos_x, pos_y, marker='2', color='k')
plt.tight_layout()
plt.show()
```
